```python
# ============================================================
# PDF RAG - Data Loading → Splitting → Embeddings → FAISS
# → Similarity Search
# ============================================================

# 1. Install required packages
!pip install -q -U langchain-community pypdf langchain-huggingface sentence-transformers faiss-cpu


# 2. Upload PDF to Google Colab
from google.colab import files

uploaded = files.upload()


# 3. Get the uploaded PDF filename automatically
import os

pdf_file = list(uploaded.keys())[0]

print("PDF loaded:", pdf_file)


# 4. Data Loading
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader(pdf_file)
text_loader = loader.load()

print("\nTotal pages:", len(text_loader))
print("\nFirst page:")
print(text_loader[0].page_content[:1000])


# 5. Document Splitting
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=200,
    chunk_overlap=20
)

final_documents = text_splitter.split_documents(text_loader)

print("\nTotal chunks:", len(final_documents))


# 6. Create Embeddings
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="all-MiniLM-L6-v2"
)

print("\nEmbedding model loaded successfully!")


# 7. Create FAISS Vector Database
from langchain_community.vectorstores import FAISS

db = FAISS.from_documents(
    final_documents,
    embeddings
)

print("FAISS vector database created successfully!")


# 8. Similarity Search
query = input("\nEnter your question for Arduino: ")

docs = db.similarity_search(query, k=4)

print("\n========== RELEVANT DOCUMENTS ==========\n")

for i, doc in enumerate(docs):
    print(f"--- Document {i + 1} ---")
    print(doc.page_content)
    print()
```
