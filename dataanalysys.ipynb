```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
  "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
  "Maths": [85, 92, 78, 90, 88],
  "Science": [90, 85, 88, 92, 95],
  "English": [88, 90, 85, 87, 92]
}
df = pd.DataFrame(data)
print(df)
```


marks = df[["Maths", "Science", "English"]].values
average_marks = np.mean(marks, axis=1)
df["Average"] = average_marks
print(df)


```python
def grade(avg):
  if avg >= 90:
    return "A"
  elif avg >= 80:
    return "B"
  elif avg >= 70:
    return "C"
  elif avg >= 60:
    return "D"
  else:
    return "F"

df["Grade"] = df["Average"].apply(grade)
```
